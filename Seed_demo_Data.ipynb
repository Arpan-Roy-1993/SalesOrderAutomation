```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F

spark = SparkSession.builder.getOrCreate()

# Base folder inside your Lakehouse Files area
base_path = "Files/Test_files"

# -------------------------
# 1. READ CSV FILES
# -------------------------
df_raw_header = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv(f"{base_path}/po_raw_header.csv")
)

df_raw_lines = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv(f"{base_path}/po_raw_lines.csv")
)

df_sales_order_status = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv(f"{base_path}/erp_sales_order_status.csv")
)

df_item_master = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv(f"{base_path}/erp_item_master.csv")
)

df_uom_conversion = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv(f"{base_path}/erp_uom_conversion.csv")
)

df_item_hold = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv(f"{base_path}/erp_item_hold.csv")
)

# -------------------------
# 2. OPTIONAL CLEANUP / CASTING
# -------------------------
df_raw_header = (
    df_raw_header
    .withColumn("document_id", F.col("document_id").cast("string"))
    .withColumn("customer_id", F.col("customer_id").cast("string"))
    .withColumn("sales_order_no", F.col("sales_order_no").cast("string"))
    .withColumn("doc_total", F.col("doc_total").cast("double"))
    .withColumn("source_file_name", F.col("source_file_name").cast("string"))
)

df_raw_lines = (
    df_raw_lines
    .withColumn("document_id", F.col("document_id").cast("string"))
    .withColumn("line_no", F.col("line_no").cast("int"))
    .withColumn("item_extracted", F.col("item_extracted").cast("string"))
    .withColumn("description_extracted", F.col("description_extracted").cast("string"))
    .withColumn("qty_extracted", F.col("qty_extracted").cast("double"))
    .withColumn("uom_extracted", F.col("uom_extracted").cast("string"))
    .withColumn("unit_price_extracted", F.col("unit_price_extracted").cast("double"))
)

df_sales_order_status = (
    df_sales_order_status
    .withColumn("sales_order_no", F.col("sales_order_no").cast("string"))
    .withColumn("sales_order_status", F.col("sales_order_status").cast("string"))
)

df_item_master = (
    df_item_master
    .withColumn("item_no", F.col("item_no").cast("string"))
    .withColumn("item_description", F.col("item_description").cast("string"))
    .withColumn("standard_uom", F.col("standard_uom").cast("string"))
    .withColumn("min_order_qty", F.col("min_order_qty").cast("double"))
)

df_uom_conversion = (
    df_uom_conversion
    .withColumn("item_no", F.col("item_no").cast("string"))
    .withColumn("from_uom", F.col("from_uom").cast("string"))
    .withColumn("to_uom", F.col("to_uom").cast("string"))
    .withColumn("conversion_factor", F.col("conversion_factor").cast("double"))
)

df_item_hold = (
    df_item_hold
    .withColumn("item_no", F.col("item_no").cast("string"))
    .withColumn("site_code", F.col("site_code").cast("string"))
    .withColumn("item_hold_flag", F.col("item_hold_flag").cast("boolean"))
    .withColumn("hold_reason", F.col("hold_reason").cast("string"))
)

# -------------------------
# 3. WRITE AS DELTA TABLES
# -------------------------
table_map = {
    "po_raw_header": df_raw_header,
    "po_raw_lines": df_raw_lines,
    "erp_sales_order_status": df_sales_order_status,
    "erp_item_master": df_item_master,
    "erp_uom_conversion": df_uom_conversion,
    "erp_item_hold": df_item_hold,
}

for table_name, df in table_map.items():
    (
        df.write
        .mode("overwrite")
        .format("delta")
        .saveAsTable(table_name)
    )

print("Tables created successfully:")
for t in table_map:
    print(f" - {t}")

print("Preview: po_raw_header")
display(df_raw_header)

print("Preview: po_raw_lines")
display(df_raw_lines)

print("Preview: erp_item_master")
display(df_item_master)
```
